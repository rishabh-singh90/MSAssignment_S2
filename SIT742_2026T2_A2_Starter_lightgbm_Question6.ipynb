### Validation and final-cutoff split objects

Create these required scaffold objects from `raw_tourism_data` using the `Date` column and the existing constants. Use a real year-month representation such as `pandas.Period(..., freq="M")`; do not filter by direct string comparison on month labels.

- `training_actual_wide_to_2023M02`: used for validation MASE denominators, validation MAPE alignment, and validation baseline generation cutoff.
- `validation_actual_wide_2023M03_2023M07`: visible validation actual/result table used for Q1-Q3 integrated evidence and Q4 validation comparison.
- `public_history_wide_to_2023M07`: public history available for the final forecasting cutoff and final MASE denominator concept; no final-evaluation actual/result values are included.


````python
FINAL_CUTOFF = "2023M07"


def month_label_to_period(month_label):
    """Convert a YYYYMmm label such as 2023M03 to a pandas monthly Period."""
    return pd.Period(str(month_label).strip().replace("M", "-"), freq="M")


def period_to_month_label(period):
    """Convert a pandas monthly Period back to a YYYYMmm label."""
    return f"{period.year}M{period.month:02d}"


raw_tourism_data_with_period = raw_tourism_data.copy()
raw_tourism_data_with_period["_month_period"] = raw_tourism_data_with_period[
    FINAL_FORECAST_DATE_COLUMN
].map(month_label_to_period)
raw_tourism_data_with_period = raw_tourism_data_with_period.sort_values("_month_period", kind="mergesort")

training_actual_wide_to_2023M02 = raw_tourism_data_with_period.loc[
    raw_tourism_data_with_period["_month_period"] <= month_label_to_period(TRAIN_END),
    FINAL_FORECAST_COLUMNS,
].reset_index(drop=True)

validation_periods = [month_label_to_period(month) for month in VALIDATION_MONTHS]
validation_actual_wide_2023M03_2023M07 = raw_tourism_data_with_period.loc[
    raw_tourism_data_with_period["_month_period"].isin(validation_periods),
    FINAL_FORECAST_COLUMNS,
].reset_index(drop=True)

public_history_wide_to_2023M07 = raw_tourism_data_with_period.loc[
    raw_tourism_data_with_period["_month_period"] <= month_label_to_period(FINAL_CUTOFF),
    FINAL_FORECAST_COLUMNS,
].reset_index(drop=True)

# Cutoff evidence: the public file contains empty rows for the final forecast
# period, so no assessment-period actual/result values are available or used.
final_period_rows = raw_tourism_data_with_period.loc[
    raw_tourism_data_with_period["_month_period"] > month_label_to_period(FINAL_CUTOFF), DESTINATION_COLUMNS
]
split_summary = pd.DataFrame(
    {
        "object": ["training_actual_wide_to_2023M02", "validation_actual_wide_2023M03_2023M07", "public_history_wide_to_2023M07"],
        "rows": [len(training_actual_wide_to_2023M02), len(validation_actual_wide_2023M03_2023M07), len(public_history_wide_to_2023M07)],
        "columns": [training_actual_wide_to_2023M02.shape[1], validation_actual_wide_2023M03_2023M07.shape[1], public_history_wide_to_2023M07.shape[1]],
        "first_month": [training_actual_wide_to_2023M02["Date"].iloc[0], validation_actual_wide_2023M03_2023M07["Date"].iloc[0], public_history_wide_to_2023M07["Date"].iloc[0]],
        "last_month": [training_actual_wide_to_2023M02["Date"].iloc[-1], validation_actual_wide_2023M03_2023M07["Date"].iloc[-1], public_history_wide_to_2023M07["Date"].iloc[-1]],
    }
)
display(split_summary)
print("Rows after the final cutoff in the public file:", len(final_period_rows),
      "| non-missing values in those rows:", int(final_period_rows.notna().sum().sum()))
assert list(validation_actual_wide_2023M03_2023M07["Date"]) == VALIDATION_MONTHS
assert final_period_rows.notna().sum().sum() == 0
display(validation_actual_wide_2023M03_2023M07)
````

### EDA 2. Data availability and missing months

The series do not all start in 1989. Coverage matters directly for Q3, because the MASE denominator uses every valid lag-1 pair in a destination's own history: a short series gets a denominator estimated from fewer, more recent movements.

````python
overview_rows = []
for destination in DESTINATION_COLUMNS:
    series = eda_history[destination]
    first_valid = series.first_valid_index()
    observed = series.loc[first_valid:]
    missing_labels = [period_to_month_label(period) for period in observed.index[observed.isna()]]
    overview_rows.append({
        "destination": destination,
        "first_observed_month": period_to_month_label(first_valid),
        "observed_months": int(observed.notna().sum()),
        "missing_months_after_start": len(missing_labels),
        "missing_month_labels": ", ".join(missing_labels) if missing_labels else "",
        "zero_values": int((observed == 0).sum()),
        "mean_2019": float(series.loc[month_label_to_period("2019M01"):month_label_to_period("2019M12")].mean()),
        "mean_2023M03_2023M07": float(series.loc[month_label_to_period("2023M03"):month_label_to_period("2023M07")].mean()),
    })
data_overview = pd.DataFrame(overview_rows)
data_overview["level_2023_vs_2019"] = data_overview["mean_2023M03_2023M07"] / data_overview["mean_2019"]
display(data_overview.round(3))

print("Destinations with missing months inside their observed span:")
display(data_overview.loc[data_overview["missing_months_after_start"] > 0,
                          ["destination", "first_observed_month", "missing_months_after_start", "missing_month_labels"]])

availability = eda_history.notna().astype(int).T
availability.columns = [period.year for period in eda_history.index]
availability_by_year = availability.T.groupby(level=0).sum().T   # observed months per year
fig, ax = plt.subplots(figsize=(14, 6))
image = ax.imshow(availability_by_year.to_numpy(), aspect="auto", cmap="Blues", vmin=0, vmax=12)
ax.set_yticks(range(len(DESTINATION_COLUMNS)))
ax.set_yticklabels(DESTINATION_COLUMNS, fontsize=8)
year_positions = range(0, availability_by_year.shape[1], 3)
ax.set_xticks(list(year_positions))
ax.set_xticklabels([availability_by_year.columns[i] for i in year_positions], rotation=45, fontsize=8)
ax.set_title("Figure 2. Data availability: observed months per calendar year (dark = 12 months observed)")
fig.colorbar(image, ax=ax, shrink=0.8, label="observed months")
plt.tight_layout()
plt.show()

print("Missing values are left as NaN throughout: Q3 excludes them from MAE, MAPE and denominator pairs,")
print("and never treats them as zero.")
````

### EDA 3. Scale and distribution across destinations

````python
pre_covid_window = eda_history.loc[month_label_to_period("2015M01"):month_label_to_period("2019M12")]
eda_descriptive_stats = pd.DataFrame({
    "min": pre_covid_window.min(),
    "median": pre_covid_window.median(),
    "mean": pre_covid_window.mean(),
    "max": pre_covid_window.max(),
    "std": pre_covid_window.std(),
})
eda_descriptive_stats["coefficient_of_variation"] = eda_descriptive_stats["std"] / eda_descriptive_stats["mean"]
eda_descriptive_stats["peak_to_trough_ratio"] = eda_descriptive_stats["max"] / eda_descriptive_stats["min"]
eda_descriptive_stats["share_of_total_demand_%"] = (
    pre_covid_window.sum() / pre_covid_window.sum().sum() * 100)
print("Descriptive statistics over the stable pre-COVID period 2015M01-2019M12:")
display(eda_descriptive_stats.sort_values("mean", ascending=False).round(3))
print(f"Largest / smallest mean monthly demand ratio: "
      f"{eda_descriptive_stats['mean'].max() / eda_descriptive_stats['mean'].min():,.0f} to 1")
print("Three largest markets account for "
      f"{eda_descriptive_stats['share_of_total_demand_%'].nlargest(3).sum():.1f}% of all pre-COVID arrivals.")
````

### EDA 4. Three demand eras

The history is not one regime. Splitting it into a pre-COVID era, a closed-border era and the reopening shows how differently a model must behave in each, and why any method that learns its level from recent months alone starts from the wrong place.

````python
def generate_naive_forecast_wide(
    historical_actual_wide,
    cutoff_label,
    forecast_months,
    required_destinations=None,
    lag=1,
    date_column="Date"
):
    # --- 0. Validate lag ---
    # raises an error if lag is not a positive integer. negative numbers,float and strings not allowed.
    if not isinstance(lag, int) or lag <= 0:
        raise ValueError("lag must be a positive integer")

    # --- 1. Destination columns ---
    all_dest_cols = [c for c in historical_actual_wide.columns if c != date_column]
    dest_cols = required_destinations if required_destinations is not None else all_dest_cols

    # --- 2. Convert Date column using helper ---
    df = historical_actual_wide.copy()
    df[date_column] = df[date_column].apply(month_label_to_period_example)
    df = df.sort_values(date_column)

    cutoff_period = month_label_to_period_example(cutoff_label)

    # --- 3. Handle duplicate dates---
    # If multiple rows exist for the same month, and destination columns are also duplicate, drop one row
    # If multiple rows exist for the same month, but destination columns are different,→ ADD them

    clean_rows = []

    for period, group in df.groupby(date_column):
        values = group[dest_cols]

        # If all rows identical → keep one
        if values.nunique().sum() == 0:
            row = values.iloc[0].copy()
        else:
            # If rows differ → sum them
            row = values.sum()

        clean_rows.append({date_column: period, **row.to_dict()})

    df = pd.DataFrame(clean_rows).sort_values(date_column)

    cutoff_period = month_label_to_period_example(cutoff_label)

    # --- 4. Actual lookup (only up to cutoff) ---
    # create a lookup table with rows upto the cutoff label. cutoff label row is included.
    actual_lookup = {
        row[date_column]: {dest: row[dest] for dest in dest_cols}
        for _, row in df[df[date_column] <= cutoff_period].iterrows()
    }

    # --- 5. Forecast lookup (recursive storage) ---
    # Create an empty dictionary which will store the forecasted values
    forecast_lookup = {}

    # --- 6. Recursive lag rule ---
    def get_value(period, dest):
        # Actual available
        if period in actual_lookup:
            return actual_lookup[period][dest]

        # Already forecast
        if period in forecast_lookup:
            return forecast_lookup[period][dest]

        # Need recursive forecast
        source_period = period - lag

        # No source available → NaN
        # gives Nan if the lag jumps before the dataset begins or recursive forecasting cannot be used.
        if (source_period not in actual_lookup) and (source_period not in forecast_lookup):
            return float("nan")

        return get_value(source_period, dest)

    # --- 7. Generate forecasts in supplied order ---
    # An empty list to store values
    forecast_rows = []

    for m in forecast_months:
        period = month_label_to_period_example(m)
        row = {dest: get_value(period, dest) for dest in dest_cols}
        forecast_lookup[period] = row
        forecast_rows.append({date_column: period, **row})

    # --- 8. Build final DataFrame ---
    out = pd.DataFrame(forecast_rows)

    # Convert Period back to "YYYYMmm" using helper
    out[date_column] = out[date_column].apply(period_to_month_label_example)

    # Ensure correct column order
    return out[[date_column] + dest_cols]
````

````python
def validate_forecast_actual_wide(
    forecast_wide_df,
    actual_wide_df=None,
    required_months=None,
    required_destinations=None,
    date_column="Date",
):
    """Return the fixed Q2 audit dictionary for wide-table validation.

    Populate every key in Q2_AUDIT_FIELDS. Forecast-only mode uses None for
    can_align and every actual-related field. Forecast-vs-actual mode subsets
    actual_wide_df to required_months before reporting actual_row_count and
    checking structural alignment.
    """
    # 1. Base Setup, Structural Extraction and Parameter Inference
    f_dates = forecast_wide_df[date_column].tolist() if date_column in forecast_wide_df.columns else []
    f_cols = [c for c in forecast_wide_df.columns if c != date_column]

    req_months = required_months if required_months is not None else sorted(list(set(f_dates)))
    req_dests = required_destinations if required_destinations is not None else f_cols

    # Pre-populate fixed-key dictionary with exact required keys and operational defaults
    audit = {
        "is_valid": False,
        "can_align": None if actual_wide_df is None else False,
        "forecast_row_count": len(forecast_wide_df),
        "actual_row_count": None,
        "expected_row_count": len(required_months) if required_months is not None else None,
        "missing_months_in_forecast": [m for m in req_months if m not in f_dates],
        "missing_months_in_actual": None if actual_wide_df is None else [],
        "extra_months_in_forecast": [m for m in f_dates if m not in req_months],
        "extra_months_in_actual_source": None if actual_wide_df is None else [],
        "missing_destinations_in_forecast": [d for d in req_dests if d not in f_cols],
        "missing_destinations_in_actual": None if actual_wide_df is None else [],
        "extra_columns_in_forecast": [c for c in f_cols if c not in req_dests],
        "duplicate_forecast_dates": len(f_dates) - len(set(f_dates)) if len(f_dates) > 0 else 0,
        "duplicate_actual_dates": None if actual_wide_df is None else 0,
        "missing_forecast_value_count": 0,
        "missing_actual_value_count": None if actual_wide_df is None else 0,
        "nonnumeric_forecast_value_count": 0,
        "nonnumeric_actual_value_count": None if actual_wide_df is None else 0,
        "nonfinite_forecast_value_count": 0,
        "nonfinite_actual_value_count": None if actual_wide_df is None else 0,
    }

    # 2. Check Forecast Matrix Data Value Quality
    f_matrix_cols = [d for d in req_dests if d in f_cols]
    if f_matrix_cols:
        f_sub = forecast_wide_df[f_matrix_cols]
        audit["missing_forecast_value_count"] = int(f_sub.isna().sum().sum())

        non_null_mask = f_sub.notna()
        for col in f_matrix_cols:
            vals = f_sub[col][non_null_mask[col]]
            numeric_mask = pd.to_numeric(vals, errors='coerce').notna()
            audit["nonnumeric_forecast_value_count"] += int((~numeric_mask).sum())

            valid_nums = vals[numeric_mask]
            audit["nonfinite_forecast_value_count"] += int((~np.isfinite(valid_nums.astype(float))).sum())

    # 3. Forecast-vs-Actual Alignment Mode Pipeline Execution
    if actual_wide_df is not None:
        a_dates_raw = actual_wide_df[date_column].tolist() if date_column in actual_wide_df.columns else []
        a_cols = [c for c in actual_wide_df.columns if c != date_column]

        # Profile un-subsetted actual baseline details
        audit["extra_months_in_actual_source"] = [m for m in a_dates_raw if m not in req_months]
        if len(a_dates_raw) > 0:
            audit["duplicate_actual_dates"] = len(a_dates_raw) - len(set(a_dates_raw))

        # Target evaluation temporal alignment: subset frame safely
        a_sub_time = actual_wide_df[actual_wide_df[date_column].isin(req_months)] if date_column in actual_wide_df.columns else actual_wide_df.copy()
        audit["actual_row_count"] = len(a_sub_time)

        a_dates_sub = a_sub_time[date_column].tolist() if date_column in a_sub_time.columns else []
        audit["missing_months_in_actual"] = [m for m in req_months if m not in a_dates_sub]
        audit["missing_destinations_in_actual"] = [d for d in req_dests if d not in a_cols]

        # Profile Value Quality across actual target matrix elements
        a_matrix_cols = [d for d in req_dests if d in a_cols]
        if a_matrix_cols:
            a_sub = a_sub_time[a_matrix_cols]
            audit["missing_actual_value_count"] = int(a_sub.isna().sum().sum())

            non_null_mask_a = a_sub.notna()
            for col in a_matrix_cols:
                vals_a = a_sub[col][non_null_mask_a[col]]
                numeric_mask_a = pd.to_numeric(vals_a, errors='coerce').notna()
                audit["nonnumeric_actual_value_count"] += int((~numeric_mask_a).sum())

                valid_nums_a = vals_a[numeric_mask_a]
                audit["nonfinite_actual_value_count"] += int((~np.isfinite(valid_nums_a.astype(float))).sum())

        # Determine if matrices are capable of structural mathematical alignment
        months_aligned = (set(f_dates) == set(req_months)) and (set(a_dates_sub) == set(req_months))
        dests_aligned = (len(audit["missing_destinations_in_forecast"]) == 0) and (len(audit["missing_destinations_in_actual"]) == 0)
        no_duplicates = (audit["duplicate_forecast_dates"] == 0) and (audit["duplicate_actual_dates"] == 0)

        if months_aligned and dests_aligned and no_duplicates:
            audit["can_align"] = True

    # 4. Global Invariant Validation Check Rules
    structural_ok = (
        len(audit["missing_months_in_forecast"]) == 0 and
        len(audit["missing_destinations_in_forecast"]) == 0 and
        audit["duplicate_forecast_dates"] == 0 and
        (audit["forecast_row_count"] == audit["expected_row_count"] if audit["expected_row_count"] is not None else True)
    )
    values_ok = (
        audit["missing_forecast_value_count"] == 0 and
        audit["nonnumeric_forecast_value_count"] == 0 and
        audit["nonfinite_forecast_value_count"] == 0
    )

    if actual_wide_df is None:
        audit["is_valid"] = bool(structural_ok and values_ok)
    else:
        actual_ok = (
            audit["can_align"] is True and
            audit["missing_actual_value_count"] == 0 and
            audit["nonnumeric_actual_value_count"] == 0 and
            audit["nonfinite_actual_value_count"] == 0
        )
        audit["is_valid"] = bool(structural_ok and values_ok and actual_ok)

    return audit
````

````python
def _a2_to_period(month_label):
    """Convert a YYYYMmm month label (or a Period) to a monthly pandas Period."""
    if isinstance(month_label, pd.Period):
        return month_label.asfreq("M")
    label = str(month_label).strip()
    try:
        return pd.Period(label.replace("M", "-"), freq="M")
    except Exception as exc:
        raise ValueError(f"Month label {month_label!r} is not a valid YYYYMmm label such as 2023M03.") from exc

def _a2_to_label(period):
    """Convert a monthly pandas Period to a YYYYMmm label."""
    return f"{period.year}M{period.month:02d}"

def _a2_try_period(month_label):
    try:
        return _a2_to_period(month_label)
    except ValueError:
        return pd.NaT


def evaluate_forecast_wide(
    forecast_wide_df,
    actual_wide_df,
    training_actual_wide_df,
    start_month,
    end_month,
    required_destinations=None,
    naive_lag=1,
    date_column="Date",
):
    """Return (destination_metrics, aggregate_metrics).

    destination_metrics is a DataFrame using DESTINATION_METRIC_COLUMNS.
    aggregate_metrics is a dictionary using AGGREGATE_METRIC_FIELDS.
    Sort training_actual_wide_df chronologically before computing all valid
    denominator pairs. If required_destinations is None, evaluate all forecast
    columns except date_column; otherwise evaluate exactly those destinations.
    """
    if isinstance(naive_lag, (bool, np.bool_)) or not isinstance(naive_lag, (int, np.integer)) or naive_lag < 1:
        raise ValueError(f"naive_lag must be a positive integer; received {naive_lag!r}.")
    naive_lag = int(naive_lag)
    for name, frame in [("forecast_wide_df", forecast_wide_df), ("actual_wide_df", actual_wide_df),
                        ("training_actual_wide_df", training_actual_wide_df)]:
        if date_column not in frame.columns:
            raise KeyError(f"{name} has no {date_column!r} column.")

    if required_destinations is None:
        destinations = [column for column in forecast_wide_df.columns if column != date_column]
    else:
        destinations = list(required_destinations)

    start_period = _a2_to_period(start_month)
    end_period = _a2_to_period(end_month)
    if end_period < start_period:
        raise ValueError("end_month must not be earlier than start_month.")
    evaluation_periods = pd.period_range(start_period, end_period, freq="M")

    def with_periods(frame):
        work = frame.copy()
        work["_period"] = work[date_column].map(_a2_try_period)
        return work.loc[work["_period"].notna()]

    def align_to_evaluation_months(frame):
        # Keep the inclusive evaluation window, keep the first row for any
        # duplicated month, and reindex so forecast and actual share one Date axis.
        work = with_periods(frame)
        work = work.loc[(work["_period"] >= start_period) & (work["_period"] <= end_period)]
        work = work.drop_duplicates("_period", keep="first").set_index("_period")
        work.index = pd.PeriodIndex(work.index, freq="M")
        return work.reindex(evaluation_periods)

    aligned_forecast = align_to_evaluation_months(forecast_wide_df)
    aligned_actual = align_to_evaluation_months(actual_wide_df)

    # Training history for the MASE denominator: sorted chronologically, only
    # months before the evaluation window (no validation/assessment actuals),
    # and placed on a complete monthly calendar so lagged pairs are true
    # calendar lags even if a month row is absent.
    training = with_periods(training_actual_wide_df)
    training = training.loc[training["_period"] < start_period]
    training = training.sort_values("_period", kind="mergesort").drop_duplicates("_period", keep="last")
    training = training.set_index("_period")
    if len(training):
        training.index = pd.PeriodIndex(training.index, freq="M")
        training = training.reindex(pd.period_range(training.index.min(), training.index.max(), freq="M"))

    def numeric_array(frame, column, length):
        if column not in frame.columns:
            return np.full(length, np.nan)
        return pd.to_numeric(frame[column], errors="coerce").to_numpy(dtype=float)

    destination_metric_rows = []
    for destination in destinations:
        forecast = numeric_array(aligned_forecast, destination, len(evaluation_periods))
        actual = numeric_array(aligned_actual, destination, len(evaluation_periods))
        training_actual = numeric_array(training, destination, len(training))

        # MAE / MASE numerator: rows where both values are numeric and finite.
        valid_error_rows = np.isfinite(actual) & np.isfinite(forecast)
        n = int(valid_error_rows.sum())
        absolute_errors = np.abs(actual[valid_error_rows] - forecast[valid_error_rows])
        mae = float(absolute_errors.mean()) if n > 0 else np.nan

        # MAPE: valid rows with nonzero actual values only (no division by zero).
        valid_mape_rows = valid_error_rows & (actual != 0)
        if valid_mape_rows.any():
            mape = float(np.mean(np.abs(actual[valid_mape_rows] - forecast[valid_mape_rows]) / np.abs(actual[valid_mape_rows])) * 100)
        else:
            mape = np.nan

        # MASE denominator: all valid lagged pairs in the cutoff-safe training history.
        if len(training_actual) > naive_lag:
            left = training_actual[naive_lag:]
            right = training_actual[:-naive_lag]
            valid_pairs = np.isfinite(left) & np.isfinite(right)
            differences = np.abs(left[valid_pairs] - right[valid_pairs])
        else:
            differences = np.array([], dtype=float)
        n_denominator_pairs = int(len(differences))
        denominator = float(differences.mean()) if n_denominator_pairs > 0 else np.nan

        denominator_ok = n_denominator_pairs > 0 and np.isfinite(denominator) and denominator > 0
        mase_available = bool(denominator_ok and n > 0 and np.isfinite(mae))
        mase = float(mae / denominator) if mase_available else np.nan

        destination_metric_rows.append({
            "destination": destination,
            "n": n,
            "mae": mae,
            "mase": mase,
            "mape": mape,
            "denominator": denominator,
            "n_denominator_pairs": n_denominator_pairs,
            "mase_available": mase_available,
            "denominator_warning": bool(not denominator_ok),
        })

    destination_metrics = pd.DataFrame(destination_metric_rows, columns=DESTINATION_METRIC_COLUMNS)
    destination_metrics = destination_metrics.astype({
        "n": int, "mae": float, "mase": float, "mape": float, "denominator": float,
        "n_denominator_pairs": int, "mase_available": bool, "denominator_warning": bool,
    })

    # Unweighted summaries of destination-level values, computed unrounded.
    available_mase = destination_metrics.loc[destination_metrics["mase_available"], "mase"].dropna()
    available_mape = destination_metrics["mape"].dropna()
    aggregate_metrics = {
        "n_destinations": int(len(destination_metrics)),
        "mean_mase": float(available_mase.mean()) if len(available_mase) else np.nan,
        "median_mase": float(available_mase.median()) if len(available_mase) else np.nan,
        "mean_mape": float(available_mape.mean()) if len(available_mape) else np.nan,
    }
    return destination_metrics, aggregate_metrics
````

````python
tourism_series_long = raw_tourism_data.copy()

tourism_series_long["Date"] = tourism_series_long["Date"].map(
    month_label_to_period
)

tourism_series_long = tourism_series_long.melt(
    id_vars=["Date"],
    value_vars=DESTINATION_COLUMNS,
    var_name="destination",
    value_name="demand",
)

tourism_series_long = tourism_series_long.sort_values(
    ["destination", "Date"]
).reset_index(drop=True)

tourism_series_long.head()
````

````python
LGBM_LAGS = [1, 2, 3, 6, 12]


def prepare_destination_series(destination):
    """
    Preserve the complete monthly calendar for one destination.
    Missing demand values remain NaN.
    """
    destination_data = tourism_series_long.loc[
        tourism_series_long["destination"] == destination,
        ["Date", "demand"],
    ].copy()

    destination_data = destination_data.sort_values("Date")
    destination_data["Date"] = pd.PeriodIndex(
        destination_data["Date"],
        freq="M",
    )

    destination_series = (
        destination_data
        .drop_duplicates(subset=["Date"])
        .set_index("Date")["demand"]
        .astype(float)
    )

    complete_index = pd.period_range(
        start=destination_series.index.min(),
        end=destination_series.index.max(),
        freq="M",
    )

    return destination_series.reindex(complete_index)


def create_lgbm_features(series):
    """
    Create time-series features while preserving monthly continuity.
    The target is the current value of the series.
    """
    series = series.astype(float)
    index = series.index

    features = pd.DataFrame(index=index)

    # Calendar and trend features
    features["year"] = index.year
    features["month"] = index.month
    features["month_sin"] = np.sin(2 * np.pi * index.month / 12)
    features["month_cos"] = np.cos(2 * np.pi * index.month / 12)
    features["trend"] = np.arange(len(index))

    # COVID period indicator.
    # This is known from the calendar and does not use demand values.
    # Use YYYY-MM, not YYYYMM03.
    covid_start = pd.Period("2020-03", freq="M")
    covid_end = pd.Period("2022-02", freq="M")
    recovery_start = pd.Period("2022-03", freq="M")

    features["covid_period"] = (
        (index >= covid_start) &
        (index <= covid_end)
    ).astype(int)

    features["post_covid_recovery"] = (
        index >= recovery_start
    ).astype(int)

    # Lag features
    for lag in LGBM_LAGS:
        features[f"lag_{lag}"] = series.shift(lag)

    # Prediction target
    features["target"] = series

    return features
````

````python
LGBM_PARAMETER_GRID = {
    "n_estimators": [100, 200],
    "learning_rate": [0.03, 0.05],
    "num_leaves": [7, 15],
    "max_depth": [-1, 5],
    "min_child_samples": [10, 20],
}


def fit_lgbm_destination(
    destination,
    cutoff_period,
    forecast_periods=None,
):
    """
    Fit one LightGBM model for one destination.

    Only data up to cutoff_period is used.
    Future actual values are never used as features or targets.
    """
    cutoff_period = pd.Period(cutoff_period, freq="M")

    series = prepare_destination_series(destination)
    features = create_lgbm_features(series)

    # Restrict training data to the modelling cutoff.
    training_data = features.loc[
        features.index <= cutoff_period
    ].copy()

    # A target must be observed.
    # Missing lag features are retained because LightGBM supports NaN features.
    training_data = training_data.dropna(subset=["target"])

    feature_columns = [
        column for column in training_data.columns
        if column != "target"
    ]

    X_train = training_data[feature_columns]
    y_train = training_data["target"]

    if len(X_train) < 30:
        raise ValueError(
            f"{destination} has only {len(X_train)} usable rows. "
            "Not enough observations for reliable LightGBM training."
        )

    # Use time-ordered cross-validation.
    n_splits = min(4, len(X_train) // 20)

    if n_splits < 2:
        raise ValueError(
            f"Not enough observations for TimeSeriesSplit: {destination}"
        )

    time_cv = TimeSeriesSplit(n_splits=n_splits)

    base_model = LGBMRegressor(
        objective="regression",
        random_state=42,
        n_jobs=-1,
        verbosity=-1,
        reg_alpha=0.1,
        reg_lambda=0.1,
    )

    grid_search = GridSearchCV(
        estimator=base_model,
        param_grid=LGBM_PARAMETER_GRID,
        scoring="neg_mean_absolute_error",
        cv=time_cv,
        n_jobs=-1,
        refit=True,
    )

    grid_search.fit(X_train, y_train)

    best_model = grid_search.best_estimator_

    def recursive_forecast(periods):
        """
        Forecast future months recursively.

        A prediction for one month is added to history and can then
        be used as a lag for the following month.
        """
        periods = pd.PeriodIndex(periods, freq="M")
        history = series.copy()

        predictions = []

        for period in periods:
            # Create a temporary history including previous predictions.
            temporary_features = create_lgbm_features(history)

            row = temporary_features.loc[[period], feature_columns]

            prediction = float(best_model.predict(row)[0])

            # Tourism demand cannot be negative.
            prediction = max(0.0, prediction)

            predictions.append(prediction)
            history.loc[period] = prediction

        return pd.Series(predictions, index=periods)

    final_forecast = None

    if forecast_periods is not None:
        final_forecast = recursive_forecast(forecast_periods)

    return {
        "destination": destination,
        "model": best_model,
        "grid_search": grid_search,
        "best_params": grid_search.best_params_,
        "cv_mae": -grid_search.best_score_,
        "training_rows": len(X_train),
        "final_forecast": final_forecast,
    }
````

````python
def month_label_to_period(month_label):
    if isinstance(month_label, pd.Period):
        return month_label.asfreq("M")

    text = str(month_label).strip()

    # Convert labels such as 2023M03 to 2023-03.
    if "M" in text:
        year, month = text.split("M")
        text = f"{year}-{month}"

    return pd.Period(text, freq="M")
````

#### Q4 step 1: shared modelling building blocks

Three of the six candidate models share one idea, so the pieces are defined once, before any model uses them.

- `_history_to_cutoff(...)` is the **cutoff guard**: every model calls it first, so no row after its cutoff can ever be read.
- `pre_covid_reference_profile(...)` is the **seasonal reference**: the mean of each calendar month over 2017M08-2019M07, the last two pre-COVID years. That window ends long before either cutoff and avoids the late-2019 distortions in Hong Kong China (social unrest) and Taiwan China (suspension of individual travel permits).
- `recovery_ratio_table(...)` is the **recovery ratio**: `demand ÷ reference for the same calendar month`. A ratio of 0.32 means demand is 32% of its pre-COVID norm for that month. Ratios are formed from 2022M01 onwards.

`modelling_data_or_feature_table` records the resulting model-ready state for all 20 destinations at both cutoffs: the recent ratios, their 3-month mean, the fitted slope over the last 6 months, and the pre-COVID reference level.

````python
PRE_COVID_REFERENCE_WINDOW = ("2017M08", "2019M07")
RECOVERY_RATIO_START = "2022M01"
# Design constraint for the ratio models: a forecast ratio may not exceed 100% of the
# pre-COVID reference, unless the destination is already above it at the cutoff.
RATIO_CAP = 1.0


def _history_to_cutoff(history_wide, cutoff_label, date_column="Date"):
    """Cutoff guard: return the history indexed by monthly Period, rows <= cutoff only."""
    cutoff_period = _a2_to_period(cutoff_label)
    work = history_wide.copy()
    work["_period"] = work[date_column].map(_a2_to_period)
    work = work.loc[work["_period"] <= cutoff_period]
    work = work.sort_values("_period", kind="mergesort").drop_duplicates("_period", keep="last").set_index("_period")
    work.index = pd.PeriodIndex(work.index, freq="M")
    destinations = [column for column in history_wide.columns if column != date_column]
    return work[destinations].apply(pd.to_numeric, errors="coerce").astype(float), cutoff_period


def pre_covid_reference_profile(history, window=PRE_COVID_REFERENCE_WINDOW):
    """Mean demand for each calendar month (1-12) inside the pre-COVID reference window."""
    start, end = (_a2_to_period(label) for label in window)
    reference_rows = history.loc[(history.index >= start) & (history.index <= end)]
    return reference_rows.groupby(reference_rows.index.month).mean().reindex(range(1, 13))


def recovery_ratio_table(history, reference_profile, start_label=RECOVERY_RATIO_START):
    """actual / pre-COVID same-calendar-month reference, from start_label to the cutoff."""
    recent = history.loc[history.index >= _a2_to_period(start_label)]
    reference = reference_profile.reindex(recent.index.month)
    reference.index = recent.index
    return recent / reference.where(reference > 0)


def ratio_state(history_wide, cutoff_label, n_recent=6, date_column="Date"):
    """Inspectable model-ready summary of each destination's recovery state at a cutoff."""
    history, cutoff_period = _history_to_cutoff(history_wide, cutoff_label, date_column)
    reference_profile = pre_covid_reference_profile(history)
    ratios = recovery_ratio_table(history, reference_profile)
    recent = ratios.loc[ratios.index > cutoff_period - n_recent]

    rows = []
    for destination in history.columns:
        values = recent[destination].to_numpy(dtype=float)
        valid = np.isfinite(values)
        row = {"cutoff": _a2_to_label(cutoff_period), "destination": destination, "n_valid_ratios": int(valid.sum())}
        for offset, period in enumerate(recent.index):
            row[f"ratio_t-{len(recent.index) - 1 - offset}"] = values[offset]
        row["mean_ratio_last3"] = float(np.nanmean(values[-3:])) if valid[-3:].any() else np.nan
        row["ratio_slope_last6"] = (
            float(np.polyfit(np.arange(len(values))[valid], values[valid], 1)[0]) if valid.sum() >= 2 else np.nan
        )
        row["last_observed_ratio"] = float(values[valid][-1]) if valid.any() else np.nan
        row["reference_annual_mean"] = float(reference_profile[destination].mean())
        series = history[destination].dropna()
        row["last_observed_value"] = float(series.iloc[-1]) if len(series) else np.nan
        rows.append(row)
    return pd.DataFrame(rows), reference_profile


validation_state, validation_reference_profile = ratio_state(training_actual_wide_to_2023M02, TRAIN_END)
final_state, final_reference_profile = ratio_state(public_history_wide_to_2023M07, FINAL_CUTOFF)
modelling_data_or_feature_table = pd.concat([validation_state, final_state], ignore_index=True)

# Leakage evidence: corrupting every value after the cutoff leaves the modelling state unchanged.
tampered_history = raw_tourism_data.copy()
tampered_history.loc[tampered_history["Date"].map(_a2_to_period) > _a2_to_period(TRAIN_END), DESTINATION_COLUMNS] = 1e9
tampered_state, _ = ratio_state(tampered_history, TRAIN_END)
pd.testing.assert_frame_equal(tampered_state, validation_state)
print("Leakage check passed: the 2023M02 modelling state does not change when post-cutoff data are altered.")

with pd.option_context("display.max_rows", 50):
    display(modelling_data_or_feature_table.round(4))
reference_display = final_reference_profile.copy()
reference_display.index.name = "calendar_month"
print("Pre-COVID reference profile (mean of 2017M08-2019M07 by calendar month):")
display(reference_display.round(0))
````

#### Q4 step 2: the six candidate models

All six share one interface — `(history table, cutoff label, forecast months)` returning a wide forecast table — so they can be swapped, backtested and audited identically. Each filters its input to the cutoff before reading any value, and each falls back to a lag-1 naive forecast for any destination whose fit fails.

| # | Model | Family | Tunable setting |
|---|---|---|---|
| 1 | `naive_lag1` | Required baseline (Q1, lag 1) | none |
| 2 | `naive_lag12` | Required baseline (Q1, lag 12) | none |
| 3 | `recovery_ratio_flat` | Recovery ratio held flat x seasonal reference | averaging window |
| 4 | `sarimax_log_demand` | Seasonal ARIMA on `log1p(demand)` | (p,d,q) and seasonal (P,D,Q)[12] |
| 5 | `sarimax_recovery_ratio` | Seasonal ARIMA on the recovery ratio x seasonal reference | (p,d,q), no seasonal term needed |
| 6 | `prophet_recovery_ratio` | Prophet trend on the recovery ratio x seasonal reference | changepoint prior scale |

Models 4 and 5 use the same estimator on different representations of the data, which isolates the effect of the recovery-ratio representation from the effect of the estimator. Model 5 needs no seasonal term because dividing by the pre-COVID monthly profile has already removed the seasonality; the seasonality is restored when the forecast ratio is multiplied back by that profile.

`prophet` is not part of the base scientific stack, so its import is guarded: if it is missing, that model is dropped from the comparison and the notebook still runs. Install it with `pip install prophet` if needed.

````python
import logging

from statsmodels.tsa.statespace.sarimax import SARIMAX


def _silence_prophet_logging():
    """Prophet's Stan backend logs one INFO line per fit; keep the notebook output readable."""
    for name in ("cmdstanpy", "prophet"):
        noisy_logger = logging.getLogger(name)
        noisy_logger.setLevel(logging.CRITICAL)
        noisy_logger.handlers = []
        noisy_logger.propagate = False
        noisy_logger.disabled = True


try:
    import prophet
    from prophet import Prophet
    _silence_prophet_logging()
    PROPHET_AVAILABLE = True
    PROPHET_VERSION = prophet.__version__
except Exception:
    PROPHET_AVAILABLE = False
    PROPHET_VERSION = None

SARIMAX_HISTORY_START = "2012M01"


def _forecast_setup(history_wide, cutoff_label, forecast_months, required_destinations, date_column):
    """Shared preparation: destinations, cutoff-filtered history, horizons and the naive fallback."""
    destinations = ([c for c in history_wide.columns if c != date_column]
                    if required_destinations is None else list(required_destinations))
    history, cutoff_period = _history_to_cutoff(history_wide, cutoff_label, date_column)
    periods = [_a2_to_period(month) for month in forecast_months]
    horizons = [max((period - cutoff_period).n, 1) for period in periods]
    fallback = generate_naive_forecast_wide(
        history_wide, cutoff_label, forecast_months, destinations, lag=1, date_column=date_column
    )
    return destinations, history, cutoff_period, periods, horizons, fallback


def _assemble(output, destinations, periods, date_column):
    forecast = pd.DataFrame(output, columns=destinations)
    forecast.insert(0, date_column, [_a2_to_label(period) for period in periods])
    return forecast


def recovery_ratio_flat_forecast_wide(
    history_wide, cutoff_label, forecast_months, required_destinations=None, window=3, date_column="Date"
):
    """Model 3: hold the mean of the last `window` recovery ratios flat across the horizon."""
    destinations, history, _, periods, _, fallback = _forecast_setup(
        history_wide, cutoff_label, forecast_months, required_destinations, date_column)
    reference_profile = pre_covid_reference_profile(history)
    ratios = recovery_ratio_table(history, reference_profile)

    output = {}
    for destination in destinations:
        values = ratios[destination].dropna().to_numpy(dtype=float)[-window:]
        level = float(values.mean()) if len(values) else np.nan
        forecasts = [level * reference_profile.loc[period.month, destination] for period in periods]
        output[destination] = ([float(value) for value in forecasts] if np.isfinite(forecasts).all()
                               else fallback[destination].astype(float).tolist())
    return _assemble(output, destinations, periods, date_column)


def sarimax_log_demand_forecast_wide(
    history_wide, cutoff_label, forecast_months, required_destinations=None,
    order=(1, 1, 1), seasonal_order=(0, 1, 1, 12), date_column="Date",
):
    """Model 4: seasonal ARIMA fitted directly to log1p(demand)."""
    destinations, history, _, periods, horizons, fallback = _forecast_setup(
        history_wide, cutoff_label, forecast_months, required_destinations, date_column)
    history = history.loc[history.index >= _a2_to_period(SARIMAX_HISTORY_START)]

    output = {}
    for destination in destinations:
        series = history[destination]
        try:
            series = series.loc[series.first_valid_index():].interpolate()
            with warnings.catch_warnings():
                warnings.simplefilter("ignore")
                fitted = SARIMAX(
                    np.log1p(series.to_numpy(dtype=float)), order=order, seasonal_order=seasonal_order,
                    enforce_stationarity=False, enforce_invertibility=False,
                ).fit(disp=False, maxiter=100)
            path = np.clip(np.expm1(fitted.forecast(max(horizons))), 0, None)
            values = [float(path[horizon - 1]) for horizon in horizons]
            if not np.isfinite(values).all():
                raise ValueError("nonfinite SARIMAX forecast")
            output[destination] = values
        except Exception:
            output[destination] = fallback[destination].astype(float).tolist()
    return _assemble(output, destinations, periods, date_column)


def sarimax_recovery_ratio_forecast_wide(
    history_wide, cutoff_label, forecast_months, required_destinations=None,
    order=(1, 1, 0), date_column="Date",
):
    """Model 5: SARIMAX on the recovery ratio, multiplied back by the seasonal reference."""
    destinations, history, _, periods, horizons, fallback = _forecast_setup(
        history_wide, cutoff_label, forecast_months, required_destinations, date_column)
    reference_profile = pre_covid_reference_profile(history)
    ratios = recovery_ratio_table(history, reference_profile)

    output = {}
    for destination in destinations:
        series = ratios[destination].dropna()
        try:
            if len(series) < 6:
                raise ValueError("too few recovery ratios")
            with warnings.catch_warnings():
                warnings.simplefilter("ignore")
                fitted = SARIMAX(
                    series.to_numpy(dtype=float), order=order, trend="c",
                    enforce_stationarity=False, enforce_invertibility=False,
                ).fit(disp=False, maxiter=100)
            cap = max(RATIO_CAP, float(series.iloc[-1]))
            path = np.clip(fitted.forecast(max(horizons)), 0, cap)
            values = [float(path[horizon - 1] * reference_profile.loc[period.month, destination])
                      for horizon, period in zip(horizons, periods)]
            if not np.isfinite(values).all():
                raise ValueError("nonfinite SARIMAX forecast")
            output[destination] = values
        except Exception:
            output[destination] = fallback[destination].astype(float).tolist()
    return _assemble(output, destinations, periods, date_column)


def prophet_recovery_ratio_forecast_wide(
    history_wide, cutoff_label, forecast_months, required_destinations=None,
    changepoint_prior_scale=0.1, date_column="Date",
):
    """Model 6: Prophet trend on the recovery ratio, multiplied back by the seasonal reference."""
    destinations, history, _, periods, _, fallback = _forecast_setup(
        history_wide, cutoff_label, forecast_months, required_destinations, date_column)
    if not PROPHET_AVAILABLE:
        return fallback
    _silence_prophet_logging()
    reference_profile = pre_covid_reference_profile(history)
    ratios = recovery_ratio_table(history, reference_profile)
    future = pd.DataFrame({"ds": pd.PeriodIndex(periods, freq="M").to_timestamp()})

    output = {}
    for destination in destinations:
        series = ratios[destination].dropna()
        try:
            if len(series) < 6:
                raise ValueError("too few recovery ratios")
            training = pd.DataFrame({"ds": series.index.to_timestamp(), "y": series.to_numpy(dtype=float)})
            with warnings.catch_warnings():
                warnings.simplefilter("ignore")
                model = Prophet(
                    yearly_seasonality=False, weekly_seasonality=False, daily_seasonality=False,
                    changepoint_prior_scale=changepoint_prior_scale, uncertainty_samples=0,
                )
                model.fit(training)
                path = model.predict(future)["yhat"].to_numpy(dtype=float)
            cap = max(RATIO_CAP, float(series.iloc[-1]))
            values = [float(min(max(path[i], 0.0), cap) * reference_profile.loc[period.month, destination])
                      for i, period in enumerate(periods)]
            if not np.isfinite(values).all():
                raise ValueError("nonfinite Prophet forecast")
            output[destination] = values
        except Exception:
            output[destination] = fallback[destination].astype(float).tolist()
    return _assemble(output, destinations, periods, date_column)


print("Model functions defined. prophet available:", PROPHET_AVAILABLE,
      ("(v" + PROPHET_VERSION + ")") if PROPHET_AVAILABLE else "(prophet model will be skipped)")
````

````python
ROLLING_ORIGINS = ["2023M02", "2023M03", "2023M04", "2023M05"]


def history_up_to(cutoff_label):
    """Public history truncated to the given cutoff, ready to hand to a model."""
    return public_history_wide_to_2023M07.loc[
        public_history_wide_to_2023M07["Date"].map(_a2_to_period) <= _a2_to_period(cutoff_label)
    ].reset_index(drop=True)


def months_after(cutoff_label, last_label=VALIDATION_END):
    cutoff_period = _a2_to_period(cutoff_label)
    return [_a2_to_label(cutoff_period + h) for h in range(1, 13) if cutoff_period + h <= _a2_to_period(last_label)]


def evaluate_at_origin(forecast_fn, origin):
    """Fit at one origin on data up to that origin only, then score with Q3."""
    history = history_up_to(origin)
    months = months_after(origin)
    forecast = forecast_fn(history, origin, months)
    _, aggregate = evaluate_forecast_wide(
        forecast, validation_actual_wide_2023M03_2023M07, history,
        months[0], months[-1], REQUIRED_DESTINATIONS, naive_lag=1,
    )
    return months, aggregate


def backtest_scores(forecast_fn):
    """Official-origin and rolling-origin summaries for one model configuration."""
    per_origin = [evaluate_at_origin(forecast_fn, origin)[1] for origin in ROLLING_ORIGINS]
    return {
        "official_mean_mase": per_origin[0]["mean_mase"],
        "official_median_mase": per_origin[0]["median_mase"],
        "official_mean_mape": per_origin[0]["mean_mape"],
        "rolling_mean_mase": float(np.mean([a["mean_mase"] for a in per_origin])),
        "rolling_median_mase": float(np.mean([a["median_mase"] for a in per_origin])),
        "rolling_mean_mape": float(np.mean([a["mean_mape"] for a in per_origin])),
    }
````

#### Q4 step 3: hyperparameter grid search

Three of the six models have a setting that must be chosen. Rather than asserting values, they are selected here, **before** any model is used, by a rule fixed in advance:

> For each model, take the setting with the lowest **rolling-origin mean MASE** across the four origins.

Rolling origins are used rather than the single official origin because four origins are less likely to reward a setting that happens to suit the reopening month. The grid is deliberately small and coarse (18 settings in total); the naive baselines have nothing to tune.

Because the settings are chosen on the same validation window that is later reported, the reported validation measures are mildly optimistic. The rolling-origin results and the flatness of the grid are the evidence that this is not severe.

````python
GRID_SPECS = {
    "recovery_ratio_flat": [
        {"window": window} for window in [2, 3, 4, 5, 6]
    ],
    "sarimax_log_demand": [
        {"order": order, "seasonal_order": seasonal_order}
        for order in [(1, 1, 1), (0, 1, 1), (2, 1, 0)]
        for seasonal_order in [(0, 1, 1, 12), (1, 1, 0, 12)]
    ],
    "sarimax_recovery_ratio": [
        {"order": order} for order in [(1, 1, 0), (0, 1, 1), (1, 0, 0), (1, 1, 1)]
    ],
    "prophet_recovery_ratio": [
        {"changepoint_prior_scale": scale} for scale in [0.05, 0.1, 0.5]
    ],
}
TUNABLE_MODEL_FUNCTIONS = {
    "recovery_ratio_flat": recovery_ratio_flat_forecast_wide,
    "sarimax_log_demand": sarimax_log_demand_forecast_wide,
    "sarimax_recovery_ratio": sarimax_recovery_ratio_forecast_wide,
    "prophet_recovery_ratio": prophet_recovery_ratio_forecast_wide,
}
TUNING_CRITERION = "rolling_mean_mase"

grid_rows = []
for model_label, settings in GRID_SPECS.items():
    if model_label == "prophet_recovery_ratio" and not PROPHET_AVAILABLE:
        continue
    model_function = TUNABLE_MODEL_FUNCTIONS[model_label]
    for setting in settings:
        forecast_fn = (
            lambda history, cutoff, months, fn=model_function, setting=setting:
            fn(history, cutoff, months, REQUIRED_DESTINATIONS, **setting)
        )
        grid_rows.append({
            "model_label": model_label,
            "setting": ", ".join(f"{key}={value}" for key, value in setting.items()),
            "setting_values": setting,
            **backtest_scores(forecast_fn),
        })

hyperparameter_grid_results = pd.DataFrame(grid_rows)
best_setting_index = hyperparameter_grid_results.groupby("model_label")[TUNING_CRITERION].idxmin()
hyperparameter_grid_results["selected_setting"] = hyperparameter_grid_results.index.isin(best_setting_index)

TUNED_MODEL_PARAMS = {
    row["model_label"]: row["setting_values"]
    for _, row in hyperparameter_grid_results.loc[best_setting_index].iterrows()
}

print(f"Grid search: {len(hyperparameter_grid_results)} settings x {len(ROLLING_ORIGINS)} origins, "
      f"chosen by lowest {TUNING_CRITERION}.")
with pd.option_context("display.max_rows", 40):
    display(
        hyperparameter_grid_results.drop(columns="setting_values")
        .sort_values(["model_label", TUNING_CRITERION]).round(3)
    )
print("Tuned settings carried forward:")
for model_label, settings in TUNED_MODEL_PARAMS.items():
    print(f"  {model_label}: {settings}")
````

#### Q4 step 6: official validation (cutoff 2023M02, months 2023M03-2023M07)

Each model is given **only** `training_actual_wide_to_2023M02`. Every forecast table is audited with the Q2 function against the visible validation actuals, then scored with the Q3 function using the official naive lag-1 MASE denominator.
