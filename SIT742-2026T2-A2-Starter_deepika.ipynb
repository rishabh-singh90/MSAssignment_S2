````python
def generate_naive_forecast_wide(
    historical_actual_wide,
    cutoff_label,
    forecast_months,
    required_destinations=None,
    lag=1,
    date_column="Date"
):
    # --- 0. Validate lag ---
    # raises an error if lag is not a positive integer. negative numbers,float and strings not allowed.
    if not isinstance(lag, int) or lag <= 0:
        raise ValueError("lag must be a positive integer")
        
    # --- 1. Destination columns ---
    all_dest_cols = [c for c in historical_actual_wide.columns if c != date_column]
    dest_cols = required_destinations if required_destinations is not None else all_dest_cols

    # --- 2. Convert Date column using helper ---
    df = historical_actual_wide.copy()
    df[date_column] = df[date_column].apply(month_label_to_period_example)
    df = df.sort_values(date_column)

    cutoff_period = month_label_to_period_example(cutoff_label)

    # --- 3. Handle duplicate dates--- 
    # If multiple rows exist for the same month, and destination columns are also duplicate, drop one row
    # If multiple rows exist for the same month, but destination columns are different,→ ADD them 
    
    clean_rows = []

    for period, group in df.groupby(date_column):
        values = group[dest_cols]

        # If all rows identical → keep one
        if values.nunique().sum() == 0:
            row = values.iloc[0].copy()
        else:
            # If rows differ → sum them
            row = values.sum()

        clean_rows.append({date_column: period, **row.to_dict()})

    df = pd.DataFrame(clean_rows).sort_values(date_column)

    cutoff_period = month_label_to_period_example(cutoff_label)

    # --- 4. Actual lookup (only up to cutoff) ---
    # create a lookup table with rows upto the cutoff label. cutoff label row is included.
    actual_lookup = {
        row[date_column]: {dest: row[dest] for dest in dest_cols}
        for _, row in df[df[date_column] <= cutoff_period].iterrows()
    }

    # --- 5. Forecast lookup (recursive storage) ---
    # Create an empty dictionary which will store the forecasted values
    forecast_lookup = {}

    # --- 6. Recursive lag rule ---
    def get_value(period, dest):
        # Actual available
        if period in actual_lookup:
            return actual_lookup[period][dest]

        # Already forecast
        if period in forecast_lookup:
            return forecast_lookup[period][dest]

        # Need recursive forecast
        source_period = period - lag

        # No source available → NaN
        # gives Nan if the lag jumps before the dataset begins or recursive forecasting cannot be used.
        if (source_period not in actual_lookup) and (source_period not in forecast_lookup):
            return float("nan")

        return get_value(source_period, dest)

    # --- 7. Generate forecasts in supplied order ---
    # An empty list to store values
    forecast_rows = []

    for m in forecast_months:
        period = month_label_to_period_example(m)
        row = {dest: get_value(period, dest) for dest in dest_cols}
        forecast_lookup[period] = row
        forecast_rows.append({date_column: period, **row})

    # --- 8. Build final DataFrame ---
    out = pd.DataFrame(forecast_rows)

    # Convert Period back to "YYYYMmm" using helper
    out[date_column] = out[date_column].apply(period_to_month_label_example)

    # Ensure correct column order
    return out[[date_column] + dest_cols]
````

````python
def validate_forecast_actual_wide(
    forecast_wide_df,
    actual_wide_df=None,
    required_months=None,
    required_destinations=None,
    date_column="Date",
):
    """Return the fixed Q2 audit dictionary for wide-table validation.

    Populate every key in Q2_AUDIT_FIELDS. Forecast-only mode uses None for
    can_align and every actual-related field. Forecast-vs-actual mode subsets
    actual_wide_df to required_months before reporting actual_row_count and
    checking structural alignment.
    """
    # 1. Base Setup, Structural Extraction and Parameter Inference
    f_dates = forecast_wide_df[date_column].tolist() if date_column in forecast_wide_df.columns else []
    f_cols = [c for c in forecast_wide_df.columns if c != date_column]

    req_months = required_months if required_months is not None else sorted(list(set(f_dates)))
    req_dests = required_destinations if required_destinations is not None else f_cols

    # Pre-populate fixed-key dictionary with exact required keys and operational defaults
    audit = {
        "is_valid": False,
        "can_align": None if actual_wide_df is None else False,
        "forecast_row_count": len(forecast_wide_df),
        "actual_row_count": None,
        "expected_row_count": len(required_months) if required_months is not None else None,
        "missing_months_in_forecast": [m for m in req_months if m not in f_dates],
        "missing_months_in_actual": None if actual_wide_df is None else [],
        "extra_months_in_forecast": [m for m in f_dates if m not in req_months],
        "extra_months_in_actual_source": None if actual_wide_df is None else [],
        "missing_destinations_in_forecast": [d for d in req_dests if d not in f_cols],
        "missing_destinations_in_actual": None if actual_wide_df is None else [],
        "extra_columns_in_forecast": [c for c in f_cols if c not in req_dests],
        "duplicate_forecast_dates": len(f_dates) - len(set(f_dates)) if len(f_dates) > 0 else 0,
        "duplicate_actual_dates": None if actual_wide_df is None else 0,
        "missing_forecast_value_count": 0,
        "missing_actual_value_count": None if actual_wide_df is None else 0,
        "nonnumeric_forecast_value_count": 0,
        "nonnumeric_actual_value_count": None if actual_wide_df is None else 0,
        "nonfinite_forecast_value_count": 0,
        "nonfinite_actual_value_count": None if actual_wide_df is None else 0,
    }

    # 2. Check Forecast Matrix Data Value Quality
    f_matrix_cols = [d for d in req_dests if d in f_cols]
    if f_matrix_cols:
        f_sub = forecast_wide_df[f_matrix_cols]
        audit["missing_forecast_value_count"] = int(f_sub.isna().sum().sum())

        non_null_mask = f_sub.notna()
        for col in f_matrix_cols:
            vals = f_sub[col][non_null_mask[col]]
            numeric_mask = pd.to_numeric(vals, errors='coerce').notna()
            audit["nonnumeric_forecast_value_count"] += int((~numeric_mask).sum())

            valid_nums = vals[numeric_mask]
            audit["nonfinite_forecast_value_count"] += int((~np.isfinite(valid_nums.astype(float))).sum())

    # 3. Forecast-vs-Actual Alignment Mode Pipeline Execution
    if actual_wide_df is not None:
        a_dates_raw = actual_wide_df[date_column].tolist() if date_column in actual_wide_df.columns else []
        a_cols = [c for c in actual_wide_df.columns if c != date_column]

        # Profile un-subsetted actual baseline details
        audit["extra_months_in_actual_source"] = [m for m in a_dates_raw if m not in req_months]
        if len(a_dates_raw) > 0:
            audit["duplicate_actual_dates"] = len(a_dates_raw) - len(set(a_dates_raw))

        # Target evaluation temporal alignment: subset frame safely
        a_sub_time = actual_wide_df[actual_wide_df[date_column].isin(req_months)] if date_column in actual_wide_df.columns else actual_wide_df.copy()
        audit["actual_row_count"] = len(a_sub_time)

        a_dates_sub = a_sub_time[date_column].tolist() if date_column in a_sub_time.columns else []
        audit["missing_months_in_actual"] = [m for m in req_months if m not in a_dates_sub]
        audit["missing_destinations_in_actual"] = [d for d in req_dests if d not in a_cols]

        # Profile Value Quality across actual target matrix elements
        a_matrix_cols = [d for d in req_dests if d in a_cols]
        if a_matrix_cols:
            a_sub = a_sub_time[a_matrix_cols]
            audit["missing_actual_value_count"] = int(a_sub.isna().sum().sum())

            non_null_mask_a = a_sub.notna()
            for col in a_matrix_cols:
                vals_a = a_sub[col][non_null_mask_a[col]]
                numeric_mask_a = pd.to_numeric(vals_a, errors='coerce').notna()
                audit["nonnumeric_actual_value_count"] += int((~numeric_mask_a).sum())

                valid_nums_a = vals_a[numeric_mask_a]
                audit["nonfinite_actual_value_count"] += int((~np.isfinite(valid_nums_a.astype(float))).sum())

        # Determine if matrices are capable of structural mathematical alignment
        months_aligned = (set(f_dates) == set(req_months)) and (set(a_dates_sub) == set(req_months))
        dests_aligned = (len(audit["missing_destinations_in_forecast"]) == 0) and (len(audit["missing_destinations_in_actual"]) == 0)
        no_duplicates = (audit["duplicate_forecast_dates"] == 0) and (audit["duplicate_actual_dates"] == 0)

        if months_aligned and dests_aligned and no_duplicates:
            audit["can_align"] = True

    # 4. Global Invariant Validation Check Rules
    structural_ok = (
        len(audit["missing_months_in_forecast"]) == 0 and
        len(audit["missing_destinations_in_forecast"]) == 0 and
        audit["duplicate_forecast_dates"] == 0 and
        (audit["forecast_row_count"] == audit["expected_row_count"] if audit["expected_row_count"] is not None else True)
    )
    values_ok = (
        audit["missing_forecast_value_count"] == 0 and
        audit["nonnumeric_forecast_value_count"] == 0 and
        audit["nonfinite_forecast_value_count"] == 0
    )

    if actual_wide_df is None:
        audit["is_valid"] = bool(structural_ok and values_ok)
    else:
        actual_ok = (
            audit["can_align"] is True and
            audit["missing_actual_value_count"] == 0 and
            audit["nonnumeric_actual_value_count"] == 0 and
            audit["nonfinite_actual_value_count"] == 0
        )
        audit["is_valid"] = bool(structural_ok and values_ok and actual_ok)

    return audit
````

````python
def _a2_to_period(month_label):
    """Convert a YYYYMmm month label (or a Period) to a monthly pandas Period."""
    if isinstance(month_label, pd.Period):
        return month_label.asfreq("M")
    label = str(month_label).strip()
    try:
        return pd.Period(label.replace("M", "-"), freq="M")
    except Exception as exc:
        raise ValueError(f"Month label {month_label!r} is not a valid YYYYMmm label such as 2023M03.") from exc

def _a2_try_period(month_label):
    try:
        return _a2_to_period(month_label)
    except ValueError:
        return pd.NaT


def evaluate_forecast_wide(
    forecast_wide_df,
    actual_wide_df,
    training_actual_wide_df,
    start_month,
    end_month,
    required_destinations=None,
    naive_lag=1,
    date_column="Date",
):
    """Return (destination_metrics, aggregate_metrics).

    destination_metrics is a DataFrame using DESTINATION_METRIC_COLUMNS.
    aggregate_metrics is a dictionary using AGGREGATE_METRIC_FIELDS.
    Sort training_actual_wide_df chronologically before computing all valid
    denominator pairs. If required_destinations is None, evaluate all forecast
    columns except date_column; otherwise evaluate exactly those destinations.
    """
    if isinstance(naive_lag, (bool, np.bool_)) or not isinstance(naive_lag, (int, np.integer)) or naive_lag < 1:
        raise ValueError(f"naive_lag must be a positive integer; received {naive_lag!r}.")
    naive_lag = int(naive_lag)
    for name, frame in [("forecast_wide_df", forecast_wide_df), ("actual_wide_df", actual_wide_df),
                        ("training_actual_wide_df", training_actual_wide_df)]:
        if date_column not in frame.columns:
            raise KeyError(f"{name} has no {date_column!r} column.")

    if required_destinations is None:
        destinations = [column for column in forecast_wide_df.columns if column != date_column]
    else:
        destinations = list(required_destinations)

    start_period = _a2_to_period(start_month)
    end_period = _a2_to_period(end_month)
    if end_period < start_period:
        raise ValueError("end_month must not be earlier than start_month.")
    evaluation_periods = pd.period_range(start_period, end_period, freq="M")

    def with_periods(frame):
        work = frame.copy()
        work["_period"] = work[date_column].map(_a2_try_period)
        return work.loc[work["_period"].notna()]

    def align_to_evaluation_months(frame):
        # Keep the inclusive evaluation window, keep the first row for any
        # duplicated month, and reindex so forecast and actual share one Date axis.
        work = with_periods(frame)
        work = work.loc[(work["_period"] >= start_period) & (work["_period"] <= end_period)]
        work = work.drop_duplicates("_period", keep="first").set_index("_period")
        work.index = pd.PeriodIndex(work.index, freq="M")
        return work.reindex(evaluation_periods)

    aligned_forecast = align_to_evaluation_months(forecast_wide_df)
    aligned_actual = align_to_evaluation_months(actual_wide_df)

    # Training history for the MASE denominator: sorted chronologically, only
    # months before the evaluation window (no validation/assessment actuals),
    # and placed on a complete monthly calendar so lagged pairs are true
    # calendar lags even if a month row is absent.
    training = with_periods(training_actual_wide_df)
    training = training.loc[training["_period"] < start_period]
    training = training.sort_values("_period", kind="mergesort").drop_duplicates("_period", keep="last")
    training = training.set_index("_period")
    if len(training):
        training.index = pd.PeriodIndex(training.index, freq="M")
        training = training.reindex(pd.period_range(training.index.min(), training.index.max(), freq="M"))

    def numeric_array(frame, column, length):
        if column not in frame.columns:
            return np.full(length, np.nan)
        return pd.to_numeric(frame[column], errors="coerce").to_numpy(dtype=float)

    destination_metric_rows = []
    for destination in destinations:
        forecast = numeric_array(aligned_forecast, destination, len(evaluation_periods))
        actual = numeric_array(aligned_actual, destination, len(evaluation_periods))
        training_actual = numeric_array(training, destination, len(training))

        # MAE / MASE numerator: rows where both values are numeric and finite.
        valid_error_rows = np.isfinite(actual) & np.isfinite(forecast)
        n = int(valid_error_rows.sum())
        absolute_errors = np.abs(actual[valid_error_rows] - forecast[valid_error_rows])
        mae = float(absolute_errors.mean()) if n > 0 else np.nan

        # MAPE: valid rows with nonzero actual values only (no division by zero).
        valid_mape_rows = valid_error_rows & (actual != 0)
        if valid_mape_rows.any():
            mape = float(np.mean(np.abs(actual[valid_mape_rows] - forecast[valid_mape_rows]) / np.abs(actual[valid_mape_rows])) * 100)
        else:
            mape = np.nan

        # MASE denominator: all valid lagged pairs in the cutoff-safe training history.
        if len(training_actual) > naive_lag:
            left = training_actual[naive_lag:]
            right = training_actual[:-naive_lag]
            valid_pairs = np.isfinite(left) & np.isfinite(right)
            differences = np.abs(left[valid_pairs] - right[valid_pairs])
        else:
            differences = np.array([], dtype=float)
        n_denominator_pairs = int(len(differences))
        denominator = float(differences.mean()) if n_denominator_pairs > 0 else np.nan

        denominator_ok = n_denominator_pairs > 0 and np.isfinite(denominator) and denominator > 0
        mase_available = bool(denominator_ok and n > 0 and np.isfinite(mae))
        mase = float(mae / denominator) if mase_available else np.nan

        destination_metric_rows.append({
            "destination": destination,
            "n": n,
            "mae": mae,
            "mase": mase,
            "mape": mape,
            "denominator": denominator,
            "n_denominator_pairs": n_denominator_pairs,
            "mase_available": mase_available,
            "denominator_warning": bool(not denominator_ok),
        })

    destination_metrics = pd.DataFrame(destination_metric_rows, columns=DESTINATION_METRIC_COLUMNS)
    destination_metrics = destination_metrics.astype({
        "n": int, "mae": float, "mase": float, "mape": float, "denominator": float,
        "n_denominator_pairs": int, "mase_available": bool, "denominator_warning": bool,
    })

    # Unweighted summaries of destination-level values, computed unrounded.
    available_mase = destination_metrics.loc[destination_metrics["mase_available"], "mase"].dropna()
    available_mape = destination_metrics["mape"].dropna()
    aggregate_metrics = {
        "n_destinations": int(len(destination_metrics)),
        "mean_mase": float(available_mase.mean()) if len(available_mase) else np.nan,
        "median_mase": float(available_mase.median()) if len(available_mase) else np.nan,
        "mean_mape": float(available_mape.mean()) if len(available_mape) else np.nan,
    }
    return destination_metrics, aggregate_metrics
````
